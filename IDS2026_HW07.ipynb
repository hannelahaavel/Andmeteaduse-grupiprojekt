**a. Read the dataset description. Who produced tha data? When, how, and for what purpose?**

**Answer:** The dataset was obtained from the Estonian Centre of Registers and Information Systems (RIK) open data portal. It originates from the Estonian Courts Information System (KIS), which collects and stores information about court proceedings and related documents.

The data is entered into the court information system as part of the everyday work of Estonian courts. RIK is responsible for developing and maintaining the information system and making the open data files available to the public. The Ministry of Justice and Digital Affairs is the responsible data controller, while Estonian courts are among the authorised processors of the data.

The selected file, tsiviilasi.xml, contains information about civil court proceedings and related documents. It is provided in XML format, which allows the data to be downloaded and processed using programming tools. According to the RIK file directory, the file was last updated on 5 November 2026.

The original purpose of collecting this data is to support the organisation of court work, manage court proceedings, collect statistics, analyse court decisions and make court decisions publicly available. The data can also be reused for research and data analysis.  In our project, we aim to reuse this data to investigate whether available information about civil court proceedings can help estimate how long a proceeding might take.

**b. Name a plausible stakeholder, their goals, and the concrete decision the analysis would support. Who benefits, who could be harmed?**

**Answer:** A plausible stakeholder would be the management of the Estonian court system. Their goal is to better understand how long different civil court proceedings take and what factors may influence their duration.

The analysis could support decisions about workload planning and resource allocation. For example, if certain types of proceedings usually take longer, courts could take this into account when planning their work.

The main beneficiaries would be courts, court employees and citizens, who could benefit from better planning and more realistic expectations about proceeding times. However, inaccurate predictions could create misleading expectations or lead to unfair comparisons between courts and judges.

The dataset consists of 103351 rows and 15 columns.

The columns describe different attributes of civil court proceedings and documents, including the proceeding number, case type, proceeding registration date, proceeding type, document number, document type, document date, document status, document creator, court, court chamber, and ECLI identifier.

Each row represents a document record associated with a civil court proceeding. A single proceeding may containmultiple rows because several documents or court decisions can be created within the same case.

The original data is provided in XML format and was imported into a tabular structure using the Python Pandas library for further analysis.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Dokumentide liikide jaotus
df["DokumendiLiik"].value_counts().head(10).plot(kind="bar")

plt.title("Distribution of document types")
plt.xlabel("Document type")
plt.ylabel("Number of documents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Dokumentide arv aastate lõikes
dates = pd.to_datetime(df["DokumendiKuupaev"], errors="coerce")
year_counts = dates.dt.year.value_counts().sort_index()

year_counts.plot(kind="line", marker="o")

plt.title("Number of documents by year")
plt.xlabel("Year")
plt.ylabel("Number of documents")
plt.grid()
plt.tight_layout()
plt.show()

# Dokumentide arv kohtute kaupa
df["Kohus"].value_counts().head(10).plot(kind="bar")

plt.title("Number of documents by court")
plt.xlabel("Court")
plt.ylabel("Number of documents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# Menetluse liikide jaotus
df["MenetluseTuup"].value_counts().plot(kind="bar")

plt.title("Distribution of proceeding types")
plt.xlabel("Proceeding type")
plt.ylabel("Number of documents")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Proceeding types: The vast majority of documents are associated with Kohtu I astme menetlus (First-instance court proceedings, nearly 90000 documents). Appeals and other proceeding types make up only a tiny fraction of the total dataset.
