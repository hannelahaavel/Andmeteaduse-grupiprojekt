The second step of CRISP-DM is data understanding. The goal is to collect, describe, explore, and verify the quality of the data to ensure it’s suitable for the analytical or modeling tasks ahead.

```python
# IMPORTS AND DATA LOADING

# Import libraries
import pandas as pd                         # Data manipulation and analysis
import zipfile                              # Reading files inside ZIP archives
import xml.etree.ElementTree as ET          # Parsing XML files


# STEP 1: OPEN AND READ THE XML FILE

# Open the ZIP archive without extracting it
with zipfile.ZipFile("../tsiviilasi.zip") as z:
    # Open the XML file stored inside the archive
    with z.open("tsiviilasi.xml") as f:
        # Parse the XML into a tree structure
        tree = ET.parse(f)

# Get the root element (the top-level element of the XML)
root = tree.getroot()


# STEP 2: CONVERT XML RECORDS INTO PYTHON DICTIONARIES

# Create an empty list to store all court document records
records = []

# Loop through each court document in the XML
for item in root:

    # Create a dictionary for the current document
    # Initialize nested decisions as an empty list to avoid NaN values
    row = {"SisulisedLahendused": []}

    # Loop through the document's fields
    for child in item:

        # Check whether the field contains nested substantive decisions
        if child.tag == "SisulisedLahendused":

            # Preserve ALL substantive decisions as a list of dictionaries
            row["SisulisedLahendused"] = [
                {
                    # Use XML tags as keys and their text as values
                    field.tag: (field.text or "").strip()
                    for field in decision
                }
                # Loop through every substantive decision
                for decision in child.findall("item")
            ]

        else:
            # Store ordinary document-level fields directly
            row[child.tag] = (child.text or "").strip()

    # Add the completed document to the records list
    records.append(row)


# STEP 3: CREATE THE DOCUMENT-LEVEL DATAFRAME

# Convert the list of dictionaries into a DataFrame
# Each row represents one court document
df = pd.DataFrame(records)


# STEP 4: REMOVE DUPLICATE DOCUMENTS

# Remove documents with duplicate document numbers
# We previously verified that the five repeated document numbers correspond to completely identical records
# keep="first" retains the first occurrence of each document
df = df.drop_duplicates(
    subset=["DokumendiNr"],
    keep="first"
).reset_index(drop=True)

# Assign a unique ID to each remaining document
# This allows us to identify which flattened decisions originated from the same document
df["DocumentID"] = range(len(df))


# STEP 5: CONVERT DATE COLUMNS TO DATETIME

# Specify columns containing dates
date_columns = [
    "MenetluseRegistreerimiseKuupaev",
    "DokumendiKuupaev"
]

# Convert date strings into pandas datetime objects
# errors="coerce" replaces invalid or missing dates with NaT
for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")


# STEP 6: CREATE THE DECISION-LEVEL DATAFRAME

# Copy the document-level DataFrame to preserve the original
df_decisions = df.copy()

# Expand nested decisions into separate rows
# Documents with multiple decisions will appear multiple times
df_decisions = df_decisions.explode(
    "SisulisedLahendused"
).reset_index(drop=True)

# Convert each decision dictionary into separate columns
# Each dictionary key becomes a separate column
# Missing decisions become rows with NaN decision-level values
decision_columns = pd.json_normalize(
    df_decisions["SisulisedLahendused"]
)

# Remove the original nested column and attach the new columns
df_decisions = pd.concat(
    [
        df_decisions.drop(columns=["SisulisedLahendused"]),
        decision_columns
    ],
    axis=1
)
```
